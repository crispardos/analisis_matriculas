```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt

df = pd.read_excel("../raw/01_MATRICULAS_ED_SUPERIOR_ARICA_Y_PARINACOTA_2021.xlsx")
```

```python
df.head(10)
```

### Variable cualitativa: Tipo de institución

```python
#Tabla de frecuencia: TIPO DE INSTITUCION
frecuencia = df.groupby('TIPO DE INSTITUCION').size().sort_values(ascending=False)
n = df.shape[0]

tabla_tipo_inst = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n)*100, 1),
    'Fi': frecuencia.cumsum(),
    'Hi': round((frecuencia.cumsum()/n)*100, 1)
})


frec_institucion = df.groupby('NOMBRE DE INSTITUCION').size().sort_values(ascending=False)

series_inst_popular = pd.DataFrame({
    'Institución más popular' : frec_institucion})

display(tabla_tipo_inst)
display(series_inst_popular)


```

```python
#Gráfico de barras: TIPO DE INSTITUCION
categorias = tabla_tipo_inst.index
f = tabla_tipo_inst['fi']

fig, ax = plt.subplots(figsize=(7,5))
barras = ax.bar(categorias, f, color='steelblue', edgecolor='black')
ax.set_title('Matrículas según tipo de institución,en Arica y Parinacota 2021 ')
ax.set_xlabel('Tipo de institución')
ax.set_ylabel('Número de matrículas')
ax.bar_label(barras)
plt.xticks(rotation=45, ha='right')
plt.show()
```

En base al tipo de institución podemos ver que, las Universidades del consejo de rectores concentran el 55.6% de las matrículas de los alumnos (7.849) siendo la moda, seguido por los
Centros de Formación Técnica con 25.5%. Los Institutos Profesionales con un 10.1% y las
Universidades Privadas cuentan con un 8.8% de matrículas. En la región las matriculas están
fuertemente dominada por la Universidad de Tarapacá (estatal, CRUCH), por lo que se deduce que en la región de Tarapacá está dominado por las universidades CHURCH y una opción póco habitual son universidades privadas sin acreditación.

### Variable cualitativa: Jornada

Tabla de frecuencia completa con frecuencias acumuladas y moda. Las categorías se ordenan de mayor a
menor frecuencia (Diurno → Vespertino → Otro → Semipresencial).


```python
#Tabla de frecuencia JORNADA
frecuencia = df.groupby('JORNADA').size().sort_values(ascending=False)
n = df.shape[0]

tabla_jornada = pd.DataFrame({
    'fi': frecuencia,
    'hi': round((frecuencia/n)*100, 1),
    'Fi': frecuencia.cumsum(),
    'Hi': round((frecuencia.cumsum()/n)*100, 1)
})
tabla_jornada
```

```python
#Gráfico circular
categorias = tabla_jornada.index
h = tabla_jornada['hi']

fig, ax = plt.subplots(figsize=(6,4))
circular = ax.pie(h, labels=categorias, autopct='%.1f%%')
ax.set_title('Matrículas según jornada, Arica y Parinacota 2021')
plt.show()
```

Podemos ver que la moda es las jorandas diurnas para los 9.961 matrículados (70.6%). Por otro lado, la jornada vespertina es su segunda opción más popular con un 29.1%. Así siendo aún la opciones preferidas en esta región jornadas presenciales en la institución y nó online

```python
#Tabla de doble entrada: TIPO DE INSTITUCION x JORNADA
tabla_tipo_jornada = pd.crosstab(df['TIPO DE INSTITUCION'], df['JORNADA'], normalize='index').round(3)*100

x = np.arange(len(tabla_tipo_jornada))
fig, ax = plt.subplots(figsize=(8,5))
b1 = ax.bar(x - 0.2, tabla_tipo_jornada['Diurno'], width=0.4, color='seagreen', edgecolor='black', label='Diurno')
b2 = ax.bar(x + 0.2, tabla_tipo_jornada['Vespertino'], width=0.4, color='darkkhaki', edgecolor='black', label='Vespertino')
ax.set_title('Composición de la matrícula por jornada según tipo de institución, 2021')
ax.set_xlabel('Tipo de institución')
ax.set_ylabel('Porcentaje de matrículas (%)')
ax.set_xticks(x)
ax.set_xticklabels(tabla_tipo_jornada.index, rotation=45, ha='right')
ax.bar_label(b1, fmt='%.1f')
ax.bar_label(b2, fmt='%.1f')
ax.legend()
plt.show()

tabla_tipo_jornada
```

Siguiendo los resultados previos de las jornadas más pópulares, podemos ver ya un patron claro. Instituciones como un CFT o IP tienden a tener porcentajes más cercanos entre jornadas diurnas y vespertinas (lo podemos asociar a la elección más popular por si trabajan los matrículados), por otro lado sigue dominando en Universidades la jornada diurna con un 88.8% en las CRUCH.

```python
#Media y mediana de EDAD por tipo de institución
tabla_edad_inst2 = df.groupby('TIPO DE INSTITUCION')['EDAD'].agg(['mean', 'median', 'size']).round(1)
tabla_edad_inst2.columns = ['Edad media', 'Edad mediana', 'Matrículas']

x = np.arange(len(tabla_edad_inst2))
fig, ax = plt.subplots(figsize=(8,5))
b1 = ax.bar(x - 0.2, tabla_edad_inst2['Edad media'], width=0.4, color='mediumpurple',
            edgecolor='black', label='Media')
b2 = ax.bar(x + 0.2, tabla_edad_inst2['Edad mediana'], width=0.4, color='darkorange',
            edgecolor='black', label='Mediana')
ax.set_title('Edad de los matriculados según tipo de institución, 2021')
ax.set_xlabel('Tipo de institución')
ax.set_ylabel('Edad (años)')
ax.set_xticks(x)
ax.set_xticklabels(tabla_edad_inst2.index, rotation=45, ha='right')
ax.bar_label(b1, fmt='%.1f')
ax.bar_label(b2, fmt='%.1f')
ax.legend()
plt.show()

tabla_edad_inst2
```

Podemos ver que indpendiente de la institución, los rangos de edades van de los 22 a 24 en su mediana. Indicando que los matriculados en general mantienen como una de sus objetivos ir a una institución superior para poder tener una carrera.

```python
#Porcentajes por fila: RANGO EDAD X TIPO DE INSTITUCION
tabla_edad_tipo = pd.crosstab(df['RANGO EDAD'], df['TIPO DE INSTITUCION'], normalize='index').round(3)*100

x = np.arange(len(tabla_edad_tipo))
fig, ax = plt.subplots(figsize=(10,5))
b1 = ax.bar(x - 0.3, tabla_edad_tipo['Universidades CRUCH'], width=0.2, color='steelblue',
            edgecolor='black', label='Universidades CRUCH')
b2 = ax.bar(x - 0.1, tabla_edad_tipo['Centros de Formacion Tecnica'], width=0.2, color='indianred',
            edgecolor='black', label='Centros de Formacion Tecnica')
b3 = ax.bar(x + 0.1, tabla_edad_tipo['Institutos Profesionales'], width=0.2, color='seagreen',
            edgecolor='black', label='Institutos Profesionales')
b4 = ax.bar(x + 0.3, tabla_edad_tipo['Universidades Privadas'], width=0.2, color='darkorange',
            edgecolor='black', label='Universidades Privadas')

ax.set_title('Distribución de las matrículas por tipo de institución según rango de edad, 2021')
ax.set_xlabel('Rango de edad')
ax.set_ylabel('Porcentaje de matrículas (%)')
ax.set_xticks(x)
ax.set_xticklabels(tabla_edad_tipo.index, rotation=45, ha='right')
ax.bar_label(b1, fmt='%.1f')
ax.bar_label(b2, fmt='%.1f')
ax.bar_label(b3, fmt='%.1f')
ax.bar_label(b4, fmt='%.1f')
ax.legend()
plt.show()

tabla_edad_tipo
```

Aquí podemos ver una gráfica más interesante. Podemos afirmar que la edad si influye en el tipo de institución que ingresan los estudiantes si ponemos atención a los rango de edades de los matriculados. 

- Los estudíantes más jóvenes se concentran en universidades CRUCH.
- Los estudiantes a partir de los 30 años prefieren centros de formación técnica con carreras más rpápidas de poder completar

Estos datos son coherentes con que para un recien egresado de enseñanza media su opción principal es una universidad CRUCH, en cambio los CFT e IP los cuales poseen jornadas vespertinas en un mayor porcentaje ofreciendo carreras más cortas tienen como público mas populara adultos que ya trabajan o que desean obtener un título más rápido. A mayor edad hay más probabilidad de estudiar tarde en una institución técnico-profesional.
